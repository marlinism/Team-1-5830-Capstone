```python
# Build holiday and sundays from 2021 to 2030
from parking_processing.utils.free_days import build_free_days
from parking_processing.utils.s3 import s3_client, parse_s3_uri

S3_ROOT = "s3://smart-park-seattle/parking_v2/" # v2
bucket, prefix = parse_s3_uri(S3_ROOT)

df = build_free_days(2021, 2030)
local = "/tmp/free_days.csv"
df.to_csv(local, index=False)

s3 = s3_client()
key = f"{prefix}meta/free_days.csv"
s3.upload_file(local, bucket, key)
print("uploaded:", f"s3://{bucket}/{key}", "rows:", len(df))
```
