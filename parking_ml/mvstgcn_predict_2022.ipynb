```python
S3_ROOT = "s3://smart-park-seattle/parking_v2/" #v2 for preprocess 
YEAR = 2023

FEAT_TOPK = 16
TIN = 12
BATCH_SIZE = 8
```

```python
from parking_processing.inference.predict_mvstgcn import PredictCfg, run_predict_mvstgcn
import json
from parking_processing.utils.s3 import s3_client, parse_s3_uri, s3_get_text

# load splits to know which weeks to predict
s3 = s3_client()
bucket, prefix = parse_s3_uri(S3_ROOT)
splits = json.loads(s3_get_text(s3, bucket, f"{prefix}splits/year={YEAR}/splits.json"))
weeks = splits["train_weeks"] + splits["val_weeks"] + splits["test_weeks"]
```

```python
# Create prediction for y_15
cfg = PredictCfg(
    s3_root=S3_ROOT,
    year=YEAR,
    model_dir="s3://smart-park-seattle/parking_v2/models/year=2022/mvstgcn/y15/model/",
    tin=TIN,
    batch_size=BATCH_SIZE,
    feat_topk=FEAT_TOPK,
    use_topology=1,
    use_feature_graph=1,
    target="y_15",
    #overwrite=True,
)
# we are using graphs (node, topology from 2022)
run_predict_mvstgcn(cfg, weeks)
```

```python
# Create prediction for y_30

cfg = PredictCfg(
    s3_root=S3_ROOT,
    year=YEAR,
    model_dir="s3://smart-park-seattle/parking_v2/models/year=2022/mvstgcn/y30/model/",
    tin=TIN,
    batch_size=BATCH_SIZE,
    feat_topk=FEAT_TOPK,
    use_topology=1,
    use_feature_graph=1,
    target="y_30",
)

run_predict_mvstgcn(cfg, weeks)
```
